```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
```

```python
print(u, v)
```

```python
u*v
```

```python
def vector_multiply(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]
    result = 0.0
    for i in range(n):
        result += u[i] * v[i]

    return result
```

```python
vector_multiply(u, v)
```

```python
u.dot(v)
```

```python
v.dot(u)
```

```python
U = np.array([[1, 2, 3], [4, 5, 6]])
U.shape
```

```python
def matrix_multiply(U, v):
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]
    result = np.zeros(num_rows)
    for i in range(num_rows):
        result[i] = vector_multiply(U[i], v)

    return result
```

```python
matrix_multiply(U, v)
```

```python
U.dot(v)
```

### Matrix-matrix multiplication

```python
V = np.array([[5, 6, 7], [8, 9, 10]])
V.shape
```

```python
def matrix_matrix_multiply(U, V):
    assert U.shape[1] == V.shape[1]

    m = U.shape[0]


    result = np.zeros((m, m))
    for i in range(m):
        result[i] = matrix_multiply(V, U[i])

    return result
```

```python
matrix_matrix_multiply(U, V)
```

```python
U.dot(V.T)
```
