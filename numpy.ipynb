### Exercise 2: Column Centreing


```python
matrix = np.array([
    [5, 10, 15],
    [10, 20, 30],
    [15, 30, 45]
])
```

- a) Compute the mean of each column.


```python
col_mean = matrix.mean(axis=0)

col_mean
```

```python
# b) Subtract the column means from the matrix.
matrix - col_mean
```

```python
# matrix.mean(axis=0)
```

```python
### Exercise 3: Row Centreing
row_means = matrix.mean(axis=1,keepdims=True)
```

```python
matrix - row_means
```

### Exercise 5: Distance from a Point

```python
points = np.array([1, 3, 7, 10, 15])
```

```python
signed_diff = points - 8
signed_diff
```

```python
abs_signed_diff = np.abs(signed_diff)
abs_signed_diff
```

```python
closest = abs_signed_diff[np.argmin(abs_signed_diff)]
closest
```
