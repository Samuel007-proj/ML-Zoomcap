```python
customers['Full_Name'] = customers['Full_Name'].str.strip()

customers.Full_Name
```

```python
customers['Email'] = customers['Email'].str.strip().str.lower()
```

```python
customers["Full_Name"] = customers["Full_Name"].str.strip().str.title()
```

```python
customers["City"] = customers["City"].str.strip().str.title()
```

```python
customers['Full_Name'] = (
    customers['Full_Name']
    .str.strip()
    .str.replace(r'\s+', " ", regex=True)
    .str.title()
)
```

```python
customers.info()
```

```python
customers.dtypes[customers.dtypes == 'object'].index
```

```python
customers['Plan_Type'] = (
    customers['Plan_Type']
    .str.strip()
    .str.lower()
    .str.capitalize()
)
```

```python
customers
```

```python
customers['Age'] = pd.to_numeric(customers['Age'], errors='coerce')

customers
```

```python
customers["Customer_ID"] = pd.to_numeric(
    customers["Customer_ID"],
    errors="coerce"
).astype("Int64")
```

```python
customers['Signup_Date'] = pd.to_datetime(
    customers['Signup_Date'],
    errors='coerce',
    format="%Y-%m-%d",
)

customers
```

```python
customers['Signup_Date'].dtype
```

```python
customers["Signup_Year"] = customers["Signup_Date"].dt.year
customers["Signup_Month"] = customers["Signup_Date"].dt.month
customers["Signup_Day"] = customers["Signup_Date"].dt.day
```

```python
customers
```

```python
customers["Signup_Month_Name"] = customers["Signup_Date"].dt.month_name()
```

```python
customers[customers.duplicated(subset=['Email'], keep=False)]
```

```python
customers = customers.drop_duplicates(subset=["Email"], keep="first")
```

```python
customers["Plan_Type"] = customers["Plan_Type"].astype("category")
customers["City"] = customers["City"].astype("category")
```

```python
customers["Age_Group"] = pd.cut(
    customers["Age"],
    bins=[0, 17, 30, 45, 120],
    labels=["Under 18", "18-30", "31-45", "46+"]
)
```
